```python
import pytumblr
import os
from dotenv import load_dotenv
```

```python
load_dotenv("../../.env")

client = pytumblr.TumblrRestClient(
    consumer_key=os.getenv("TUMBLR_CONSUMER_KEY"),
    consumer_secret=os.getenv("TUMBLR_CONSUMER_SECRET")
    )
```

```python
posts = client.tagged(
        "python", 
        limit=5,
        
    )
for post in posts:
    print(post["blog_name"])
    print(post["post_url"])
    print(post["date"])
    print(post["summary"])
    print()
```
