```python
transactions = [
    {"country": "SG", "category": "Skincare", "amount": 50},
    {"country": "SG", "category": "Makeup", "amount": 80},
    {"country": "MY", "category": "Skincare", "amount": 40},
    {"country": "SG", "category": "Skincare", "amount": 30},
    {"country": "MY", "category": "Makeup", "amount": 100},
    {"country": "MY", "category": "Skincare", "amount": 60},
]
```

```python
# {
#     ("SG", "Skincare"): 80,
#     ("SG", "Makeup"): 80,
#     ("MY", "Skincare"): 100,
#     ("MY", "Makeup"): 100
# }

from collections import defaultdict

def calculate_sales(transactions):
    sales = defaultdict(float)

    for t in transactions:
        key = (t["country"], t["category"])

        if key in sales:
            sales[key] += t["amount"]
        else:
            sales[key] = t["amount"]

    sorted_sales = sorted(sales.items(), key=lambda x: x[1], reverse=True)

    return sorted_sales
```

```python
print(calculate_sales(transactions))
```
