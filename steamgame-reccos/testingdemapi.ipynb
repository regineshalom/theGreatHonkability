```python
import os
from pathlib import Path
from steam_web_api import Steam
from dotenv import load_dotenv

print(Path().resolve())

load_dotenv()

KEY = os.environ.get("steamkeyapi")
STEAM_ID = os.environ.get("steam-id")
STEAM_USERNAME = os.environ.get("steam-username")
steam = Steam(KEY)
print(steam)
```

```python
steam.users.get_user_details(steam_id=STEAM_ID)
```

```python
load_dotenv()

steamid = steam.users.search_user(STEAM_USERNAME).get('player').get('steamid')
print(steamid)
# steam.users.get_user_recently_played_games(steamid)
ownedGames = steam.users.get_owned_games(steamid)
print(ownedGames)
# 
```

```python
import requests

# url = "https://partner.steam-api.com/IPlayerService/GetOwnedGames/v1/"
url = "https://api.steampowered.com/IPlayerService/GetOwnedGames/v1/"

params = {
  "key": KEY,
  "steamid": STEAM_ID,
  "include_appinfo": True,
  "include_played_free_games": True
}

response = requests.get(url, params=params)
print(response.status_code)
print(response.url)
print(response.text)
# print("Status=", response.status_code)
# print("URL=", response.url)
# print("Full json")
```

```python
testing_url = "https://api.steampowered.com/ISteamUser/GetFriendList/v1/"

params = {
    "key": KEY,
    "steamid": STEAM_ID
}

response2 = requests.get(testing_url, params=params).json()
print(response2)
```

```python
testing2_url = "https://api.steampowered.com/IPlayerService/GetRecentlyPlayedGames/v1"

params = {
    "key": KEY,
    "steamid": STEAM_ID,
    "include_appinfo": 1,
    "include_played_free_games": 1,
    "format": "json",
}

response3 = requests.get(testing2_url, params=params, timeout=30)
print(response3)
print(response3.status_code)
print(response3.url)
print(response3.text)
```

```python
import requests
import json

url = "https://api.steampowered.com/IWishlistService/GetWishlist/v1/"

params = {
    "steamid": STEAM_ID,
    "include_appinfo": True
}

resp = requests.get(url, params=params, timeout=30)
resp.raise_for_status()
data = resp.json()
items = data.get("response", {}).get("items", [])
if not items:
    # Some accounts return items at top level
    items = data.get("response", {}).get("wishlist", []) or []
items
```
